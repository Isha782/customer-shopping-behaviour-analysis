```python
# 1. IMPORT LIBRARIES

import warnings

warnings.filterwarnings(
    "ignore",
    category=DeprecationWarning,
    module="jupyter_client.session"
)

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_val_score
)

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder,
    LabelEncoder
)

from sklearn.impute import SimpleImputer

from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score,
    roc_curve
)

from sklearn.preprocessing import label_binarize

sns.set_theme(style="whitegrid")

# 2. LOAD DATASET

DATA_PATH = "/content/shopping_trends.csv"

df = pd.read_csv(DATA_PATH)

print("\nDataset loaded successfully.")

print("\nDataset shape:")
print(df.shape)

print("\nFirst five records:")
display(df.head())

print("\nLast five records:")
display(df.tail())

# 3. DATASET INFORMATION

print("\nDataset information:")
df.info()

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes.to_frame("Data Type"))

print("\nStatistical summary:")
display(df.describe(include="all").T)

# 4. DATA QUALITY CHECKS

print("\nMissing values in each column:")
display(df.isnull().sum().to_frame("Missing Values"))

print("\nTotal missing values:")
print(df.isnull().sum().sum())

print("\nDuplicate records:")
print(df.duplicated().sum())

print("\nUnique values in each column:")
display(df.nunique().to_frame("Unique Values"))

# 5. DATA CLEANING

# Remove duplicate records if any
df = df.drop_duplicates().copy()

# Remove unnecessary spaces from column names
df.columns = df.columns.str.strip()

# Remove leading/trailing spaces from text columns
for column in df.select_dtypes(include="object").columns:
    df[column] = df[column].astype("string").str.strip()

# Convert numerical columns
numeric_columns = [
    "Age",
    "Purchase Amount (USD)",
    "Review Rating",
    "Previous Purchases"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

# Display cleaned dataset
print("\nCleaned dataset shape:")
print(df.shape)

print("\nMissing values after cleaning:")
display(df.isnull().sum().to_frame("Missing Values"))

print("\nCleaned dataset preview:")
display(df.head())

# 6. DEFINE DATASET VARIABLES

target_column = "Subscription Status"

if target_column not in df.columns:
    raise ValueError(
        f"Target column '{target_column}' was not found."
    )

if df[target_column].isnull().any():
    df = df.dropna(subset=[target_column]).copy()

print("\nTarget variable distribution:")
display(df[target_column].value_counts().to_frame("Count"))

print("\nTarget variable percentages:")
display(
    (df[target_column].value_counts(normalize=True) * 100)
    .round(2)
    .to_frame("Percentage")
)

# 7. EXPLORATORY DATA ANALYSIS (EDA)

print("\n================ EDA ================")


# 7.1 Age distribution

plt.figure(figsize=(9, 5))

sns.histplot(
    df["Age"].dropna(),
    bins=20,
    kde=True
)

plt.title("Distribution of Customer Age")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
plt.close()


# 7.2 Purchase amount distribution

plt.figure(figsize=(9, 5))

sns.histplot(
    df["Purchase Amount (USD)"].dropna(),
    bins=20,
    kde=True
)

plt.title("Distribution of Purchase Amount")
plt.xlabel("Purchase Amount (USD)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
plt.close()


# 7.3 Gender distribution

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Gender"
)

plt.title("Customer Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
plt.close()


# 7.4 Product category distribution

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="Category",
    order=df["Category"].value_counts().index
)

plt.title("Purchases by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Number of Purchases")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
plt.close()


# 7.5 Seasonal shopping distribution

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Season",
    order=df["Season"].value_counts().index
)

plt.title("Customer Purchases by Season")
plt.xlabel("Season")
plt.ylabel("Number of Purchases")
plt.tight_layout()
plt.show()
plt.close()


# 7.6 Subscription status distribution

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Subscription Status"
)

plt.title("Customer Subscription Status")
plt.xlabel("Subscription Status")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
plt.close()


# 7.7 Average purchase amount by category

category_purchase = (
    df.groupby("Category")["Purchase Amount (USD)"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(9, 5))

sns.barplot(
    x=category_purchase.index,
    y=category_purchase.values
)

plt.title("Average Purchase Amount by Category")
plt.xlabel("Product Category")
plt.ylabel("Average Purchase Amount (USD)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
plt.close()


# 7.8 Purchase amount by gender

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Gender",
    y="Purchase Amount (USD)"
)

plt.title("Purchase Amount by Gender")
plt.xlabel("Gender")
plt.ylabel("Purchase Amount (USD)")
plt.tight_layout()
plt.show()
plt.close()


# 7.9 Review rating distribution

plt.figure(figsize=(8, 5))

sns.histplot(
    df["Review Rating"].dropna(),
    bins=15,
    kde=True
)

plt.title("Distribution of Customer Review Ratings")
plt.xlabel("Review Rating")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
plt.close()


# 7.10 Correlation heatmap

numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(9, 6))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix of Numerical Variables")
plt.tight_layout()
plt.show()
plt.close()

# 8. DESCRIPTIVE ANALYSIS

print("\n================ DESCRIPTIVE ANALYSIS ================")

print("\nAverage purchase amount:")
print(df["Purchase Amount (USD)"].mean())

print("\nAverage customer age:")
print(df["Age"].mean())

print("\nAverage review rating:")
print(df["Review Rating"].mean())

print("\nAverage previous purchases:")
print(df["Previous Purchases"].mean())

print("\nTotal purchase amount recorded:")
print(df["Purchase Amount (USD)"].sum())


# Purchase amount by category
print("\nPurchase amount by category:")

category_summary = (
    df.groupby("Category")
    .agg(
        Number_of_Purchases=("Category", "count"),
        Average_Purchase=("Purchase Amount (USD)", "mean"),
        Total_Purchase=("Purchase Amount (USD)", "sum"),
        Average_Rating=("Review Rating", "mean")
    )
    .round(2)
    .sort_values("Total_Purchase", ascending=False)
)

display(category_summary)


# Subscription comparison
print("\nSubscription comparison:")

subscription_summary = (
    df.groupby("Subscription Status")
    .agg(
        Customer_Count=("Customer ID", "count"),
        Average_Purchase=("Purchase Amount (USD)", "mean"),
        Average_Rating=("Review Rating", "mean"),
        Average_Previous_Purchases=("Previous Purchases", "mean")
    )
    .round(2)
)

display(subscription_summary)

# 9. K-MEANS CUSTOMER SEGMENTATION

print("\n================ K-MEANS CLUSTERING ================")

# Customer ID is an identifier and should not be used
# as a clustering feature.

clustering_features = [
    "Age",
    "Purchase Amount (USD)",
    "Review Rating",
    "Previous Purchases"
]

cluster_data = df[clustering_features].copy()

# Fill missing numerical values
cluster_imputer = SimpleImputer(strategy="median")

cluster_data = pd.DataFrame(
    cluster_imputer.fit_transform(cluster_data),
    columns=clustering_features,
    index=df.index
)

# Standardise numerical variables
scaler = StandardScaler()

scaled_data = scaler.fit_transform(cluster_data)


# 9.1 Elbow method

inertia_values = []

for k in range(2, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(scaled_data)

    inertia_values.append(kmeans.inertia_)


plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 11),
    inertia_values,
    marker="o"
)

plt.title("Elbow Method for Optimal Number of Clusters")
plt.xlabel("Number of Clusters")
plt.ylabel("Within-Cluster Sum of Squares")
plt.xticks(range(2, 11))
plt.tight_layout()
plt.show()
plt.close()


# 9.2 Silhouette score comparison

silhouette_results = []

for k in range(2, 8):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(scaled_data)

    score = silhouette_score(
        scaled_data,
        labels,
        sample_size=min(2000, len(scaled_data)),
        random_state=42
    )

    silhouette_results.append({
        "Clusters": k,
        "Silhouette Score": score
    })


silhouette_df = pd.DataFrame(silhouette_results)

print("\nSilhouette scores:")
display(silhouette_df.round(4))


# Select number of clusters based on silhouette score.
# Also inspect the elbow chart before interpreting the result.

optimal_k = int(
    silhouette_df.loc[
        silhouette_df["Silhouette Score"].idxmax(),
        "Clusters"
    ]
)

print("\nSelected number of clusters:", optimal_k)


# 9.3 Fit final K-Means model

final_kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

df["Customer Segment"] = final_kmeans.fit_predict(
    scaled_data
)


# 9.4 Segment profile

segment_profile = (
    df.groupby("Customer Segment")
    .agg(
        Customer_Count=("Customer ID", "count"),
        Average_Age=("Age", "mean"),
        Average_Purchase=("Purchase Amount (USD)", "mean"),
        Average_Rating=("Review Rating", "mean"),
        Average_Previous_Purchases=("Previous Purchases", "mean")
    )
    .round(2)
)

print("\nCustomer segment profiles:")
display(segment_profile)


# 9.5 Segment size

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Customer Segment",
    order=sorted(df["Customer Segment"].unique())
)

plt.title("Number of Customers in Each Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
plt.close()


# 9.6 PCA visualisation

pca = PCA(n_components=2)

pca_data = pca.fit_transform(scaled_data)

pca_df = pd.DataFrame({
    "Principal Component 1": pca_data[:, 0],
    "Principal Component 2": pca_data[:, 1],
    "Customer Segment": df["Customer Segment"].astype(str).values
})

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=pca_df,
    x="Principal Component 1",
    y="Principal Component 2",
    hue="Customer Segment",
    palette="tab10",
    alpha=0.7
)

plt.title("Customer Segmentation Using PCA")
plt.tight_layout()
plt.show()
plt.close()

# 10. CLASSIFICATION: SUBSCRIPTION PREDICTION

print("\n================ CLASSIFICATION ================")

# Target variable
y = df[target_column].astype(str)

excluded_columns = [
    "Customer ID",
    target_column,
    "Customer Segment"
]

X = df.drop(
    columns=[
        col for col in excluded_columns
        if col in df.columns
    ]
).copy()


# Identify numerical and categorical columns

numeric_features = X.select_dtypes(
    include=np.number
).columns.tolist()

categorical_features = X.select_dtypes(
    exclude=np.number
).columns.tolist()


print("\nNumerical features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)


# Preprocessing for numerical features

numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)


# Preprocessing for categorical features

categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ]
)


# Combine preprocessing

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)


# Split data into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining dataset shape:", X_train.shape)
print("Testing dataset shape:", X_test.shape)

# 11. DECISION TREE CLASSIFIER

decision_tree = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "classifier",
            DecisionTreeClassifier(
                random_state=42,
                max_depth=8,
                min_samples_split=10,
                min_samples_leaf=5,
                class_weight="balanced"
            )
        )
    ]
)

decision_tree.fit(X_train, y_train)

dt_predictions = decision_tree.predict(X_test)

print("\nDECISION TREE RESULTS")

print("\nClassification report:")
print(
    classification_report(
        y_test,
        dt_predictions,
        zero_division=0
    )
)

dt_accuracy = accuracy_score(
    y_test,
    dt_predictions
)

dt_precision = precision_score(
    y_test,
    dt_predictions,
    pos_label="Yes",
    zero_division=0
)

dt_recall = recall_score(
    y_test,
    dt_predictions,
    pos_label="Yes",
    zero_division=0
)

dt_f1 = f1_score(
    y_test,
    dt_predictions,
    pos_label="Yes",
    zero_division=0
)


# Confusion matrix

dt_cm = confusion_matrix(
    y_test,
    dt_predictions,
    labels=["No", "Yes"]
)

plt.figure(figsize=(6, 5))

sns.heatmap(
    dt_cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["No", "Yes"],
    yticklabels=["No", "Yes"]
)

plt.title("Decision Tree Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.show()
plt.close()

# 12. RANDOM FOREST CLASSIFIER

random_forest = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "classifier",
            RandomForestClassifier(
                n_estimators=200,
                max_depth=12,
                min_samples_split=5,
                min_samples_leaf=2,
                class_weight="balanced",
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)

random_forest.fit(X_train, y_train)

rf_predictions = random_forest.predict(X_test)

print("\nRANDOM FOREST RESULTS")

print("\nClassification report:")
print(
    classification_report(
        y_test,
        rf_predictions,
        zero_division=0
    )
)

rf_accuracy = accuracy_score(
    y_test,
    rf_predictions
)

rf_precision = precision_score(
    y_test,
    rf_predictions,
    pos_label="Yes",
    zero_division=0
)

rf_recall = recall_score(
    y_test,
    rf_predictions,
    pos_label="Yes",
    zero_division=0
)

rf_f1 = f1_score(
    y_test,
    rf_predictions,
    pos_label="Yes",
    zero_division=0
)


# Confusion matrix

rf_cm = confusion_matrix(
    y_test,
    rf_predictions,
    labels=["No", "Yes"]
)

plt.figure(figsize=(6, 5))

sns.heatmap(
    rf_cm,
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=["No", "Yes"],
    yticklabels=["No", "Yes"]
)

plt.title("Random Forest Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.show()
plt.close()

# 13. MODEL PERFORMANCE COMPARISON

performance_df = pd.DataFrame({
    "Model": [
        "Decision Tree",
        "Random Forest"
    ],
    "Accuracy": [
        dt_accuracy,
        rf_accuracy
    ],
    "Precision (Yes)": [
        dt_precision,
        rf_precision
    ],
    "Recall (Yes)": [
        dt_recall,
        rf_recall
    ],
    "F1 Score (Yes)": [
        dt_f1,
        rf_f1
    ]
})

print("\nMODEL PERFORMANCE COMPARISON")

display(
    performance_df.round(4)
)


# Performance comparison chart

performance_plot = performance_df.set_index(
    "Model"
)

performance_plot.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Classification Model Performance Comparison")
plt.ylabel("Score")
plt.xlabel("Classification Model")
plt.ylim(0, 1.05)
plt.xticks(rotation=0)
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()
plt.close()

# 14. ROC CURVE AND AUC

# Positive class is Subscription Status = Yes

dt_probability = decision_tree.predict_proba(X_test)

rf_probability = random_forest.predict_proba(X_test)

dt_yes_index = list(
    decision_tree.classes_
).index("Yes")

rf_yes_index = list(
    random_forest.classes_
).index("Yes")

dt_prob_yes = dt_probability[:, dt_yes_index]

rf_prob_yes = rf_probability[:, rf_yes_index]

y_test_binary = (
    y_test == "Yes"
).astype(int)

dt_auc = roc_auc_score(
    y_test_binary,
    dt_prob_yes
)

rf_auc = roc_auc_score(
    y_test_binary,
    rf_prob_yes
)

dt_fpr, dt_tpr, _ = roc_curve(
    y_test_binary,
    dt_prob_yes
)

rf_fpr, rf_tpr, _ = roc_curve(
    y_test_binary,
    rf_prob_yes
)

plt.figure(figsize=(8, 6))

plt.plot(
    dt_fpr,
    dt_tpr,
    label=f"Decision Tree (AUC = {dt_auc:.3f})"
)

plt.plot(
    rf_fpr,
    rf_tpr,
    label=f"Random Forest (AUC = {rf_auc:.3f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.title("ROC Curve Comparison")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.tight_layout()
plt.show()
plt.close()


print("\nROC-AUC RESULTS")

print("Decision Tree AUC:", round(dt_auc, 4))
print("Random Forest AUC:", round(rf_auc, 4))

# 15. CROSS-VALIDATION

print("\n================ CROSS-VALIDATION ================")

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

dt_cv_scores = cross_val_score(
    decision_tree,
    X_train,
    y_train,
    cv=cv,
    scoring="accuracy",
    n_jobs=-1
)

rf_cv_scores = cross_val_score(
    random_forest,
    X_train,
    y_train,
    cv=cv,
    scoring="accuracy",
    n_jobs=-1
)

cv_results = pd.DataFrame({
    "Model": [
        "Decision Tree",
        "Random Forest"
    ],
    "Mean CV Accuracy": [
        dt_cv_scores.mean(),
        rf_cv_scores.mean()
    ],
    "Standard Deviation": [
        dt_cv_scores.std(),
        rf_cv_scores.std()
    ]
})

print("\nFive-fold cross-validation results:")
display(cv_results.round(4))

# 16. RANDOM FOREST FEATURE IMPORTANCE

print("\n================ FEATURE IMPORTANCE ================")

# Obtain transformed feature names

fitted_preprocessor = random_forest.named_steps[
    "preprocessor"
]

feature_names = fitted_preprocessor.get_feature_names_out()

feature_importance = (
    random_forest.named_steps["classifier"]
    .feature_importances_
)

importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": feature_importance
})

importance_df = importance_df.sort_values(
    "Importance",
    ascending=False
)

print("\nTop 15 important features:")
display(importance_df.head(15).round(4))


# Feature importance chart

plt.figure(figsize=(10, 7))

top_features = importance_df.head(15).sort_values(
    "Importance",
    ascending=True
)

plt.barh(
    top_features["Feature"],
    top_features["Importance"]
)

plt.title("Top 15 Random Forest Feature Importances")
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()
plt.close()

# 17. APRIORI ASSOCIATION RULE MINING

print("\n================ APRIORI ASSOCIATION RULES ================")

try:

    from mlxtend.frequent_patterns import (
        apriori,
        association_rules
    )

    association_columns = [
        "Gender",
        "Category",
        "Season",
        "Subscription Status",
        "Payment Method",
        "Shipping Type",
        "Discount Applied",
        "Promo Code Used",
        "Frequency of Purchases"
    ]

    association_columns = [
        col for col in association_columns
        if col in df.columns
    ]

    association_data = df[
        association_columns
    ].astype("string")

    # One-hot encode categorical attributes
    encoded_data = pd.get_dummies(
        association_data,
        prefix=association_columns,
        dtype=bool
    )

    # Find frequent itemsets
    frequent_itemsets = apriori(
        encoded_data,
        min_support=0.05,
        use_colnames=True,
        max_len=3
    )

    print("\nFrequent itemsets:")
    display(
        frequent_itemsets.sort_values(
            "support",
            ascending=False
        ).head(20)
    )

    # Generate association rules
    if len(frequent_itemsets) > 0:

        try:
            rules = association_rules(
                frequent_itemsets,
                metric="confidence",
                min_threshold=0.30
            )

        except TypeError:
            # Compatibility with versions requiring num_itemsets
            rules = association_rules(
                frequent_itemsets,
                num_itemsets=len(encoded_data),
                metric="confidence",
                min_threshold=0.30
            )

        if not rules.empty:

            selected_rules = rules.sort_values(
                ["lift", "confidence"],
                ascending=False
            )

            print("\nTop association rules:")

            display(
                selected_rules[
                    [
                        "antecedents",
                        "consequents",
                        "support",
                        "confidence",
                        "lift"
                    ]
                ].head(20).round(4)
            )

        else:
            print(
                "No association rules met the selected "
                "confidence threshold."
            )

    else:
        print("No frequent itemsets were identified.")

except ImportError:

    print(
        "The mlxtend library is not installed. "
        "Install it using: pip install mlxtend"
    )
```
